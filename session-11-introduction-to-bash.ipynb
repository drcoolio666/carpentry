## 1. Where Am I? -- Navigation

The first thing to learn in any new environment is how to orient yourself. In a graphical file manager, you can see your current folder and its contents at a glance. In the terminal, you ask explicit questions: *where am I?* and *what is here?*

### `pwd` -- print working directory

This tells you your current location in the filesystem. Think of it as "you are here" on a map.

````python
!pwd
````

### `ls` -- list files

On its own, `ls` shows the names of files and directories in your current location. Adding flags changes what information is displayed:

- `ls -l` -- long format (permissions, owner, size, date)
- `ls -a` -- show hidden files (those starting with `.`)
- `ls -la` -- both: long format including hidden files
- `ls -lh` -- long format with human-readable file sizes (KB, MB, GB instead of raw bytes)

````python
!ls
````

````python
!ls -la
````

````python
!ls -lh
````

### `cd` -- change directory

`cd` moves you to a different location. In a notebook, each `!` command runs in its own shell, so `!cd /tmp` has no lasting effect. We use `%cd` (a Jupyter "magic" command) instead, which changes the notebook's working directory persistently.

Key shortcuts:
- `cd ..` -- go up one level (to the parent directory)
- `cd ~` -- go to your home directory
- `cd /` -- go to the root of the filesystem
- **Absolute paths** start with `/` (e.g., `/home/jordan/data`) -- they specify the full location
- **Relative paths** do not start with `/` (e.g., `data/raw`) -- they are relative to where you currently are

````python
%cd /tmp
!pwd
````

````python
%cd ~
!pwd
````

## 3. File Permissions

Linux is a multi-user system. Every file has an owner, a group, and a set of permissions that control who can read, write, or execute it. You will encounter this most often when you write a shell script and try to run it -- Linux will refuse unless the file has "execute" permission.

### Reading `ls -l` output

When we run `ls -l`, the first column shows the permission string. For example:

```
-rwxr-xr-x  1 user group  4096 Jan 15 10:00 train.sh
```

The permission string `-rwxr-xr-x` breaks down as:

| Position | Meaning |
|----------|---------|
| `-` | File type (`-` = regular file, `d` = directory) |
| `rwx` | **Owner** permissions: read, write, execute |
| `r-x` | **Group** permissions: read, no write, execute |
| `r-x` | **Others** permissions: read, no write, execute |

A `-` in any position means that permission is not granted.

### `chmod` -- change permissions

The most common use case: you write a shell script and want to make it executable.

- `chmod +x script.sh` -- add execute permission
- `chmod -x script.sh` -- remove execute permission

````python
%%writefile /tmp/ml-project/scripts/train.sh
#!/bin/bash
echo "Starting training..."
echo "Training complete."
````

interactively we could:
```bash
cat > /tmp/ml-project/scripts/train.sh

... write your file!

```

````python
# Before: no execute permission
!ls -l /tmp/ml-project/scripts/train.sh
````

````python
# Add execute permission
!chmod +x /tmp/ml-project/scripts/train.sh

# After: note the 'x' in the permissions
!ls -l /tmp/ml-project/scripts/train.sh
````

````python
# Now we can run it
!/tmp/ml-project/scripts/train.sh
````

The `#!/bin/bash` line at the top of the script (called a "shebang") tells Linux which program to use to interpret the file. Without it, the system would not know whether the file is a Bash script, a Python script, or something else.

## 4. Environment Variables

Environment variables are named values that configure how programs behave. They are the command line's equivalent of settings -- invisible but always present. Machine learning workflows rely on them heavily.

### Viewing existing variables

A few important ones are set automatically by the system:

````python
!echo "Home directory: $HOME"
!echo "Current user:   $USER"
!echo "Shell:          $SHELL"
````

### `$PATH` -- where Linux looks for commands

`PATH` is a colon-separated list of directories. When you type a command like `python`, Linux searches each directory in `PATH` from left to right until it finds an executable with that name.

````python
# Show PATH, one directory per line for readability
!echo $PATH | tr ':' '\n'
````

### Setting your own variables

`export` makes a variable available to all programs launched from the current shell. Without `export`, the variable only exists within the shell itself.

````python
!export EXPERIMENT_NAME="baseline" && echo "Experiment: $EXPERIMENT_NAME"
````

### Why environment variables matter for ML

Many ML tools and libraries read configuration from environment variables rather than command-line arguments:

| Variable | Purpose |
|----------|---------|
| `CUDA_VISIBLE_DEVICES` | Controls which GPUs a training script can see |
| `WANDB_API_KEY` | Authenticates with Weights & Biases |
| `HF_TOKEN` | Authenticates with HuggingFace |
| `PYTHONPATH` | Adds directories to Python's import search path |
| `OMP_NUM_THREADS` | Controls the number of CPU threads for linear algebra |

For example, if you have two GPUs but want your script to use only the second one:

```bash
CUDA_VISIBLE_DEVICES=1 python train.py
```

This sets the variable for that single command without changing it globally.

### Making variables persistent

Variables set with `export` only last for the current shell session. To make them permanent, you add `export` lines to a configuration file called `~/.bashrc` (or `~/.bash_profile` on some systems). This file is executed automatically every time a new terminal opens.

We will not demonstrate this here because Kaggle and Colab environments reset between sessions, but in a real setup on your own machine or a cloud server, you would:


## 5. Pipes and Redirection -- Composing Commands

This is where Bash becomes genuinely powerful. The Unix philosophy is to build small, focused tools that each do one thing well, and then *compose* them using pipes. A pipe (`|`) takes the output of one command and feeds it as input to the next.

Think of it like a production line: each station performs one operation and passes the result along.

### Redirection: `>` and `>>`

- `command > file` -- send output to a file (overwrites the file)
- `command >> file` -- append output to a file (adds to the end)

````python
# Write output to a file
!echo "Run 1: accuracy 85%" > /tmp/ml-project/logs/results.txt

# Append to the same file
!echo "Run 2: accuracy 87%" >> /tmp/ml-project/logs/results.txt
!echo "Run 3: accuracy 91%" >> /tmp/ml-project/logs/results.txt

!cat /tmp/ml-project/logs/results.txt
````

### Pipes: `|`

A pipe connects the output of one command to the input of the next. Let us see some practical examples.

````python
# Show only the first 5 entries in a directory listing
!ls -la /usr/bin | head -5
````

````python
# Count the rows in our CSV (including header)
!cat /tmp/demo-project/data/raw/sample.csv | wc -l
````

````python
# See just the header row of a CSV
!cat /tmp/demo-project/data/raw/sample.csv | head -1
````

### `grep` -- search for patterns

`grep` filters lines that match a pattern. It is indispensable for searching through log files, code, and data.

- `grep "pattern" file` -- find lines containing "pattern"
- `grep -c "pattern" file` -- count matching lines
- `grep -i "pattern" file` -- case-insensitive search
- `grep -r "pattern" directory` -- search recursively through all files

### Command-line arguments with `sys.argv`

Real training scripts accept hyperparameters as command-line arguments. The simplest way to access them is through `sys.argv`, which is a list of all the arguments passed to the script. (Later, you will learn about `argparse`, which provides a much more robust way to handle arguments.)

````python
%%writefile /tmp/ml-project/scripts/train.py
import sys

print(f"Script name: {sys.argv[0]}")
print(f"All arguments: {sys.argv[1:]}")

# Parse simple arguments
lr = float(sys.argv[1]) if len(sys.argv) > 1 else 0.001
epochs = int(sys.argv[2]) if len(sys.argv) > 2 else 10

print(f"\nTraining configuration:")
print(f"  Learning rate: {lr}")
print(f"  Epochs: {epochs}")
print(f"\nSimulating training...")

for epoch in range(1, epochs + 1):
    fake_loss = 2.3 / epoch  # Fake decreasing loss
    print(f"  Epoch {epoch}/{epochs} - loss: {fake_loss:.4f}")

print("Done!")
````

````python
# Run with default arguments
!python /tmp/ml-project/scripts/train.py
````

````python
# Run with custom arguments
!python /tmp/ml-project/scripts/train.py 0.01 5
````

## 8. Processes and Monitoring

When you run a training job, it becomes a *process* -- a running program managed by the operating system. Knowing how to find, monitor, and stop processes is essential, especially when a training run hangs or you need to check GPU utilisation.

### Viewing running processes

`ps aux` lists all running processes. It produces a lot of output, so we pipe through `grep` to find what we care about.

````python
# Find all Python processes
!ps aux | grep python | grep -v grep
````

The `grep -v grep` at the end filters out the `grep` command itself from the results (since it also contains the word "python").

### `nvidia-smi` -- GPU monitoring

If there is one command every ML practitioner needs to know, it is `nvidia-smi`. It shows:

- Which GPU(s) are available and their model names
- GPU temperature and power usage
- Memory usage (how much of the GPU's RAM is occupied)
- Which processes are using the GPU

If a GPU is not available (as may be the case on Kaggle without GPU enabled), the command will fail -- that is normal.

````python
!nvidia-smi 2>/dev/null || echo "No GPU available in this environment. Enable GPU in Kaggle Settings or Colab Runtime to see nvidia-smi output."
````

In a terminal (not a notebook), you can monitor the GPU in real time with:

```bash
watch -n 1 nvidia-smi
```

This refreshes the `nvidia-smi` output every second. Press `Ctrl+C` to stop.

### Running processes in the background

Adding `&` to the end of a command runs it in the background, freeing your terminal for other work. Combined with `nohup`, the process continues even if you close the terminal -- essential for long training runs on remote machines.

```bash
# Run in background
python train.py &

# Run in background AND survive terminal close
nohup python train.py > output.log 2>&1 &
```

The `> output.log 2>&1` part redirects both normal output and error messages to a log file. Without this, `nohup` writes to a default file called `nohup.out`.
